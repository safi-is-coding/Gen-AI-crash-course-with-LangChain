### Streaming and Batch

Streaming is crucial for enhancing the responsiveness of applications built on LLMs. By displaying output progressively, even before a complete response is ready, streaming significantly improves user experience (UX), particularly when dealing with the latency of LLMs.

```python
model.invoke("Write me a 500 words paragrah on AI")
```

```python
for chunck in model.stream("Write me a 500 words paragrah on AI"):
    print(chunck.text, end=" ", flush=True)
```

```python
model.invoke("Why do parrots have colourful feathers")
```

```python
for chunck in model.stream("Why do parrots have colourful feathers"):
    print(chunck.text, end=" ", flush=True)
```

### Batch

LangChain handles batch processing primarily through the .batch() method available on Runnables and LLM classes, which allows developers to process multiple inputs concurrently rather than sequentially.  This approach significantly reduces latency and operational costs by minimizing the overhead of repeated API calls and parallelizing I/O-bound tasks like LLM inference.

```python
responses = model.batch(
    [
        "Why do parrots have colourful feathers?",
        "How do aeroplanes fly?",
        "What is quantum computing?"
    ]
)

for response in responses:
    print(response)
```
